```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

class RetailAnalyzer:

    def __init__(self):
        self.data=pd.DataFrame()

    def load_data(self,file):
        if file.endswith(".csv"):
            self.data=pd.read_csv(file)
            print("File loaded successfully.")
        else:
            print("Please enter a CSV file.")

    def check_data(self):
        columns=["Date","Product","Category","Price","Quantity Sold","Total Sales"]

        for col in columns:
            if col not in self.data.columns:
                print("Missing column:",col)

        print("Dataset checked.")

    def clean_data(self):
        for col in self.data.columns:
            if self.data[col].isnull().sum()>0:
                if col in ["Price","Quantity Sold","Total Sales"]:
                    self.data[col]=self.data[col].fillna(self.data[col].mean())
                else:
                    self.data[col]=self.data[col].fillna("Unknown")

        self.data["Date"]=pd.to_datetime(self.data["Date"])
        self.data["Total Sales"]=self.data["Price"]*self.data["Quantity Sold"]

        print("Data cleaned.")

    def calculate_metrics(self):
        sales=np.array(self.data["Total Sales"])

        total=np.sum(sales)
        average=np.mean(sales)

        product=self.data.groupby("Product")["Quantity Sold"].sum()
        popular=product.idxmax()

        category=self.data.groupby("Category")["Total Sales"].sum()
        best=category.idxmax()

        print("\nSales Metrics")
        print("Total Sales:",round(total,2))
        print("Average Sales:",round(average,2))
        print("Popular Product:",popular)
        print("Best Category:",best)

    def filter_category(self):
        name=input("Enter category: ")
        data=self.data[self.data["Category"].str.lower()==name.lower()]
        print(data)

    def filter_date(self):
        start=input("Enter start date(YYYY-MM-DD): ")
        end=input("Enter end date(YYYY-MM-DD): ")

        data=self.data[(self.data["Date"]>=start)&(self.data["Date"]<=end)]
        print(data)

    def display_summary(self):
        print("\nSales Summary")
        print("Rows:",len(self.data))
        print("Products:",self.data["Product"].nunique())
        print("Categories:",self.data["Category"].nunique())
        print("Total Sales:",round(self.data["Total Sales"].sum(),2))
        print("Average Sales:",round(self.data["Total Sales"].mean(),2))
        print("Quantity Sold:",self.data["Quantity Sold"].sum())

    def visualizations(self):
        print("\n1. Bar chart")
        print("2. Line graph")
        print("3. Heatmap")

        choice=input("Enter choice: ")

        if choice=="1":
            data=self.data.groupby("Product")["Total Sales"].sum()

            plt.bar(data.index,data.values)
            plt.title("Total Sales by Product")
            plt.xlabel("Product")
            plt.ylabel("Total Sales")
            plt.xticks(rotation=45)
            plt.tight_layout()
            plt.show()

        elif choice=="2":
            data=self.data.groupby("Date")["Total Sales"].sum()

            plt.plot(data.index,data.values,marker="o")
            plt.title("Sales Trend Over Time")
            plt.xlabel("Date")
            plt.ylabel("Total Sales")
            plt.xticks(rotation=45)
            plt.tight_layout()
            plt.show()

        elif choice=="3":
            data=self.data[["Price","Quantity Sold","Total Sales"]].corr()

            sns.heatmap(data,annot=True)
            plt.title("Sales Correlation")
            plt.tight_layout()
            plt.show()

        else:
            print("Invalid choice.")

def main():

    print("Retail Sales Data Analyzer")

    file=input("Enter CSV file path: ")

    analyzer=RetailAnalyzer()

    analyzer.load_data(file)

    if len(analyzer.data)>0:
        analyzer.check_data()
        analyzer.clean_data()

        while True:
            print("\n---------- MENU ----------")
            print("1. Display summary")
            print("2. Filter by category")
            print("3. Filter by date range")
            print("4. Show visualizations")
            print("5. Show cleaned data")
            print("6. Exit")

            choice=input("Enter choice: ")

            if choice=="1":
                analyzer.display_summary()
                analyzer.calculate_metrics()

            elif choice=="2":
                analyzer.filter_category()

            elif choice=="3":
                analyzer.filter_date()

            elif choice=="4":
                analyzer.visualizations()

            elif choice=="5":
                print(analyzer.data)

            elif choice=="6":
                print("Thank you for using, Your prgram ends here...")
                break

            else:
                print("Invalid choice.")

if __name__=="__main__":
    main()
```
